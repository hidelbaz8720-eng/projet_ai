```python

# Variables numériques à étudier
variables = [
    "temp",
    "hum",
    "moistureLevel",
    "totalLitres",
    "pres",
    "waterFlow"
]

# Création des histogrammes
df[variables].hist(
    figsize=(15, 10),
    bins=40,
    edgecolor="black"
)

plt.suptitle(
    "Distribution des variables numériques",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python

# Calcul de la corrélation
correlation = df[variables].corr()

# Visualisation
plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Matrice de corrélation")
plt.show()
```
